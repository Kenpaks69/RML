## 1. Overview
Analyzes demand trends across days of the week, months, and credential types.


```python
import sys
from pathlib import Path
from collections import Counter
ML_ROOT = Path('.').resolve().parent if Path('.').resolve().name == 'notebooks' else Path('.').resolve()
if str(ML_ROOT) not in sys.path:
    sys.path.insert(0, str(ML_ROOT))

from src.regquest_ml.common.paths import PROCESSED_DATA_DIR
from src.regquest_ml.data.loader import load_processed_data

records = load_processed_data(str(PROCESSED_DATA_DIR / 'cleaned_requests.csv'))
print(f'Cleaned records loaded: {len(records):,}')

cred_counts = Counter(r['primary_credential'] for r in records)
print('\nTop Credential Types:')
for c, cnt in cred_counts.most_common(7):
    print(f'  - {c:<10}: {cnt:>5} ({cnt/len(records)*100:5.1f}%)')
```

```python
daily_ts = load_processed_data(str(PROCESSED_DATA_DIR / 'daily_time_series.csv'))
weekday_totals = Counter()
weekday_counts = Counter()
for d in daily_ts:
    if d.get('is_operating_day') == '1':
        w = d['weekday_name']
        weekday_totals[w] += int(d['total_requests'])
        weekday_counts[w] += 1

print('Average Daily Requests by Weekday:')
for w in ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']:
    avg = weekday_totals[w] / max(1, weekday_counts[w])
    print(f'  - {w:<10}: {avg:.1f} requests/day')
```
