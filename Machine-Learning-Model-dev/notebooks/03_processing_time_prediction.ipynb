## 1. Overview
Calculates expected operating turnaround duration in business days.


```python
import sys
from pathlib import Path
ML_ROOT = Path('.').resolve().parent if Path('.').resolve().name == 'notebooks' else Path('.').resolve()
if str(ML_ROOT) not in sys.path:
    sys.path.insert(0, str(ML_ROOT))

from src.regquest_ml.processing_time.predict import ProcessingTimePredictor, ADMINISTRATIVE_STANDARD_DAYS

predictor = ProcessingTimePredictor()
print('Turnaround Standards Sample:')
for cred in ['TOR', 'HD', 'CAV', 'COE', 'CAR', 'DIPLOMA']:
    res = predictor.predict_one(cred, num_docs=1)
    print(f'  - {cred:<8}: {res["predicted_processing_time"]} operating days')
```

```python
print('\nTesting Multi-Document Package Lead Time Adjustment:')
single = predictor.predict_one('TOR', num_docs=1)
multi = predictor.predict_one('TOR', num_docs=2)
print(f'  - Single TOR: {single["predicted_processing_time"]} days')
print(f'  - Multi-Document TOR package: {multi["predicted_processing_time"]} days')
```
